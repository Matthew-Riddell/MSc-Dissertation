### ANOVA
Research Question

Do different ECI petitions exhibit significantly different participation rates?

#### Hypotheses:

H0

All petitions have the same mean signatures per capita

H1

At least one petition has a different mean signatures per capita

```python
from scipy.stats import f_oneway

petition_groups = []

for petition in merged_data["Petition_Name"].unique():

    group = merged_data[
        merged_data["Petition_Name"] == petition
    ]["Signatures_Per_Capita"]

    petition_groups.append(group)
```

```python
f_stat, p_value = f_oneway(*petition_groups)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

```python
if p_value < 0.05:
    print(
        "Significant difference between petitions"
    )
else:
    print(
        "No significant difference between petitions"
    )
```
