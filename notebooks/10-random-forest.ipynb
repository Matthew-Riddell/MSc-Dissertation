## Log Transform of Signatures Per Capita

### Master Dataset 
Master dataset contains some zero values for signatures per capita

```python
master["Log_Signatures_Per_Capita"] = np.log1p(
    master["Signatures_Per_Capita"]
)
```

### Successful Master Dataset

THe successful petitions contain 0 zero values for signatures per capita

```python
successful_master["Log_Signatures_Per_Capita"] = np.log(
    successful_master["Signatures_Per_Capita"]
)
```

## Random Forest 1

Model on Master Dataset

### Feature Selection

```python
X = master[
    [
        "Media_Literacy_Index",
        "Governance", 
        "Social_Capital", 
        "Personal_Freedom", 
        "Education"
    ]
]

y = master["Signatures_Per_Capita"]
```

### Checking the Data

```python
print(X.isnull().sum())
print(y.isnull().sum())
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

### Splitting data into Training and Test Sets

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=43
)
```

### Random Forest Model

```python
rf_1 = RandomForestRegressor(
    n_estimators=500,
    random_state=43,
    n_jobs=-1
)
```

### Training the model

```python
rf_1.fit(
    X_train,
    y_train
)
```

### Predictions

```python
y_train_pred = rf_1.predict(X_train)

y_test_pred = rf_1.predict(X_test)
```

### Model Evaluation

```python
train_r2 = r2_score(
    y_train,
    y_train_pred
)

test_r2 = r2_score(
    y_test,
    y_test_pred
)

print("Training R Squared:", train_r2)
print("Testing R Squared:", test_r2)
```

### RMSE

```python
train_rmse = np.sqrt(
    mean_squared_error(
        y_train,
        y_train_pred
    )
)

test_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_test_pred
    )
)

print("Training RMSE:", train_rmse)
print("Testing RMSE:", test_rmse)
```

### MAE

```python
train_mae = mean_absolute_error(
    y_train,
    y_train_pred
)

test_mae = mean_absolute_error(
    y_test,
    y_test_pred
)

print("Training MAE:", train_mae)
print("Testing MAE:", test_mae)
```

### Feature Importance

```python
feature_importance = pd.DataFrame({
    "Variable": X.columns,
    "Importance": rf_1.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print(feature_importance)
```

This shows somewhat that in the case of the Master dataset, variability is too great among non successful petitions creating too much noise for the model to adequately predict signatures per capita for Question 2

The dataset that is restricted to successful petitions is much more predictable, and the log transformed Signatures Per Capita works a lot better with the variables in each model

In the case for Question 3, it has been successfully proved that variance in signatures per capita can be explained, this is something that The Interim Report had yet to discover as it had been previosuly concluded that Socio Economic Factors didnt contribute much to the analysis
