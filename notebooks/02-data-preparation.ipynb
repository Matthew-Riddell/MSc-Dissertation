## Data Cleaning

### Cleaning Petition Datasets

### Remove total signatures rows from the dataset

checks every row to see if there is a value contained in the country column, total signatures row doesnt contain a value there so it is NaN 

it then excludes it from the dataset

```python
eci_data = eci_data[
    ~eci_data["Country"].str.contains(
        "Total number of signatories",
        case=False,
        na=False
    )
]
```

### Remove Sponsor and Metadata rows

```python
eci_data = eci_data[
    ~eci_data["Country"].str.contains(
        "Name of sponsor|Date|Description",
        case=False,
        na=False
    )
]
```

### Removing commas from numeric columns

#### Signatures column

```python
eci_data["Signatures"] = (
    eci_data["Signatures"]
    .astype(str)
    .str.replace(",", "", regex=False)
)
```

#### Threshold Column

```python
eci_data["Threshold"] = (
    eci_data["Threshold"]
    .astype(str)
    .str.replace(",", "", regex=False)
)
```

#### Percentage Column

```python
eci_data["Percentage"] = (
    eci_data["Percentage"]
    .astype(str)
    .str.replace(",", "", regex=False)
)
```

### Remove Percentage Sign from Percentage column

```python
eci_data["Percentage"] = (
    eci_data["Percentage"]
    .astype(str)
    .str.replace("%", "", regex=False)
)
```

### Convert columns into numeric 

#### Signatures column

```python
eci_data["Signatures"] = pd.to_numeric(
    eci_data["Signatures"],
    errors="coerce"
)
```

#### Threshold Column

```python
eci_data["Threshold"] = pd.to_numeric(
    eci_data["Threshold"],
    errors="coerce"
)
```

#### Percentage Column

```python
eci_data["Percentage"] = pd.to_numeric(
    eci_data["Percentage"],
    errors="coerce"
)
```

### Outlier detection

```python
print("Detecting outliers using IQR method:\n")

for col in numeric_columns:
    if col in world_selected.columns:
        Q1 = world_selected[col].quantile(0.25) # https://www.w3schools.com/Python/pandas/ref_df_quantile.asp
        Q3 = world_selected[col].quantile(0.75) # https://www.w3schools.com/Python/pandas/ref_df_quantile.asp
        IQR = Q3 - Q1
        lower_limit = Q1 - 1.5 * IQR
        upper_limit = Q3 + 1.5 * IQR
        
        outliers = world_selected[(world_selected[col] < lower_limit) | (world_selected[col] > upper_limit)]
        print(col, "has", outliers.shape[0], "outliers")
```
