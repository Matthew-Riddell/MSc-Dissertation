### Loading ECI datasets into one dataset
This combines the data from all selected petitions into one dataset while creating a petition id to differentiate 

```python
eci_dfs = []
petition_totals = []

for file_name in eci_files:
    
    file_path_eci = os.path.join(file_path, file_name)
    
    df = pd.read_csv(file_path_eci)
    
    # Add id for each petition
    df["petition_id"] = file_name.replace(".csv", "")
    
    eci_dfs.append(df)

# Combine into one dataset
eci_data = pd.concat(eci_dfs, ignore_index=True)
```

## Data Cleaning

### Remove total signatures rows from the dataset

checks every row to see if there is a value contained in the country column, total signatures row doesnt contain a value there so it is NaN 

it then excludes it from the dataset

```python
eci_data = eci_data[
    ~eci_data["Country"].str.contains(
        "Total number of signatories",
        case=False,
        na=False
    )
]
```

### Remove Sponsor and Metadata rows

```python
eci_data = eci_data[
    ~eci_data["Country"].str.contains(
        "Name of sponsor|Date|Description",
        case=False,
        na=False
    )
]
```

### Removing commas from numeric columns

#### Signatures column

```python
eci_data["Signatures"] = (
    eci_data["Signatures"]
    .astype(str)
    .str.replace(",", "", regex=False)
)
```

#### Threshold Column

```python
eci_data["Threshold"] = (
    eci_data["Threshold"]
    .astype(str)
    .str.replace(",", "", regex=False)
)
```

#### Percentage Column

```python
eci_data["Percentage"] = (
    eci_data["Percentage"]
    .astype(str)
    .str.replace(",", "", regex=False)
)
```

### Remove Percentage Sign from Percentage column

```python
eci_data["Percentage"] = (
    eci_data["Percentage"]
    .astype(str)
    .str.replace("%", "", regex=False)
)
```

### Convert columns into numeric 

#### Signatures column

```python
eci_data["Signatures"] = pd.to_numeric(
    eci_data["Signatures"],
    errors="coerce"
)
```

#### Threshold Column

```python
eci_data["Threshold"] = pd.to_numeric(
    eci_data["Threshold"],
    errors="coerce"
)
```

#### Percentage Column

```python
eci_data["Percentage"] = pd.to_numeric(
    eci_data["Percentage"],
    errors="coerce"
)
```

### Standardise country names

this is so they match the country names in the world dataset

```python
country_mapping = {
    "Czechia": "Czech Republic",
    "Ireland": "Republic of Ireland"
}

eci_data["Country"] = eci_data["Country"].replace(country_mapping)
```
