### Function to manually add the year the ECI was created based on the dataset number

```python
def get_year(number):

    if number <= 9:
        return 2024
    elif number <= 19:
        return 2023
    else:
        return 2022
```

### Loop through each individual dataset, extract data and add variables to new list

loop through each file 

dataset id determined by file name,

using file name apply the petition year (datasets were scraped in order so they are batched in numerical order)

read the dataframe to extract the total signatures row, using iloc to grab first column of all rows and finding the one that contains "Total"

if the row isn't empty, it grabs the second column and sets it as a new total_signatures variable

otherwise it makes it a NaN if no row was located

for each petition, a dictionary is created with the new variables and each dictionary is appended to a list


```python
petition_data = []

for file in eci_files:

    dataset_id = file.replace(".csv", "")
    petition_number = int(dataset_id.split("_")[-1])

    df = pd.read_csv(os.path.join(file_path, file))

    # Find the total signatures row
    total_row = df[df.iloc[:, 0].astype(str).str.contains("Total", case=False, na=False)]

    if not total_row.empty:
        value = total_row.iloc[0, 1]
        total_signatures = int(str(value).replace(",", "").strip())
    else:
        total_signatures = None

    # Store petition metadata
    petition_data.append({
        "Petition_ID": dataset_id,
        "Petition_Number": petition_number,
        "Year": get_year(petition_number),
        "Petition_Name": petition_names[dataset_id],
        "Genre": petition_genres[dataset_id],
        "Total_Signatures": total_signatures,
        "Successful": total_signatures >= 1_000_000 if total_signatures is not None else False
    })
```

### Create a new dataframe based on the petition data list

```python
petition_df = pd.DataFrame(petition_data)
```

### Sort dataframe by petition number

```python
# Sort by petition number
petition_df = petition_df.sort_values("Petition_Number").reset_index(drop=True)
```
