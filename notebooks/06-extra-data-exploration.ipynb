### Correlation Matrices

```python
print("\nPearson correlation matrix:")
pearson_corr = numeric.corr(method='pearson', numeric_only=True)
display(pearson_corr)

print("\nSpearman correlation matrix:")
spearman_corr = numeric.corr(method='spearman', numeric_only=True)
display(spearman_corr)

print("\nKendall's Tau correlation matrix:")
kendall_corr = numeric.corr(method='kendall', numeric_only=True)
display(kendall_corr)
```

```python
plt.figure(figsize=(12,10))

sns.heatmap(
    pearson_corr,
    annot=True,
    cmap="coolwarm",
    center=0
)

plt.figure(figsize=(12,10))

sns.heatmap(
    spearman_corr,
    annot=True,
    cmap="coolwarm",
    center=0
)

plt.figure(figsize=(12,10))

sns.heatmap(
    kendall_corr,
    annot=True,
    cmap="coolwarm",
    center=0
)
```

### Analysis of Media Literacy Index

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.scatter(
    media_eda["Media_Literacy_Index"],
    media_eda["Signatures_Per_Capita"]
)

plt.xlabel("Media Literacy Index")
plt.ylabel("Signatures Per Capita")
plt.title("Media Literacy Index vs Signatures Per Capita")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    media_eda["Media_Literacy_Index"],
    media_eda["Signatures_Per_Capita"]
)

plt.yscale("log")

plt.xlabel("Media Literacy Index")
plt.ylabel("Signatures Per Capita (log scale)")
plt.title("Media Literacy Index vs Signatures Per Capita")

plt.show()
```

```python
media_by_petition = (
    media_eda
    .groupby(["Petition_ID", "Petition_Name"])
    .apply(
        lambda x: x["Media_Literacy_Index"]
        .corr(
            x["Signatures_Per_Capita"],
            method="spearman"
        )
    )
    .reset_index(name="Spearman_Correlation")
)

print(media_by_petition)
```

```python
plt.figure(figsize=(12, 8))

plt.barh(
    media_by_petition["Petition_Name"],
    media_by_petition["Spearman_Correlation"]
)

plt.axvline(0, linewidth=1)

plt.xlabel("Spearman Correlation")
plt.ylabel("Petition")
plt.title(
    "Media Literacy and Signatures Per Capita "
    "by European Citizens' Initiative"
)

plt.tight_layout()
plt.show()
```

correlation between signatures per capita and media literacy varies greatly across petition

the stronger correlations appear to be among the most successful petitions but overall it varies greatly

```python
media_by_petition.sort_values(
    "Spearman_Correlation",
    ascending=False
)
```
