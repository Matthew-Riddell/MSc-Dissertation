This dataset contains a total of 27 petitions across 2022, 2023, 2024

with 27 EU member states this brings a total of 27 x 27 observations

A total of 729 rows 

19 columns comprised of variables retrived from the country level world stat data, media literacy index, petition level data and the prosperity and development dataset

The final dataset contains the following variables:

- "Petition_ID" = Unique identifier for each Initiative (string)
- "Petition_Name" = Name of the Initiative (String) Categorical
- "Country" = Name of the country (String) Categorical
- "Genre" = Genre of Initiative (String) Categorical 
- "Year" = Year of the Initiative (Int) Discreet
- "Total_Signatures" = Total signatures collected per Initiative (Int) Discreet
- "Successful" = True/False of the Initiative succesfully passed (Boolean) Discreet
- "Signatures" = Number of collected Signatures (Int) Discreet
- "Signatures_Per_Capita" = Signatures / Population per country (Float) Continuous 
- "Population" = Population of the country (int) Discreet
- "Population_Density" = Population per km squared per country (float) Continuous
- "GDP_Per_Capita" = Gross Domestic Product / Population per country (float) continuous 
- "Unemployment" = Rate of unemployment per country (float) continuous
- "Tertiary_Enrollment" = Rate of third level enrollment per country (float) continuous
- "Life_Expectancy" = Estimate average life expectancy in years (float) continous
- "Media_Literacy_Index" = Combined index that measures internet and social media use and media activity per country (float) continuous
- "Governance" = Measure of Quality of democracy, rule of law, and government effectiveness per country (float) continuous
- "Social_Capital" = Measure of Strength of personal relationships and civic engagement per country (float) continuous
- "Personal_Freedom" = Measure of Rights to speech, assembly, and individual autonomy per country (float) continuous
- "Education" = Measure of Quality and accessibility of education per country (float) continuous

### Country Participation Analysis

```python
country = (
    master
    .groupby("Country")["Signatures_Per_Capita"]
    .mean()
    .sort_values(ascending=False)
)

country.plot.bar(figsize=(14,6))
```

### Petition Analysis

```python
petition = (
    master
    .groupby("Petition_Name")["Total_Signatures"]
    .first()
    .sort_values()
)

petition.plot.bar(figsize=(14,6))
```

### Genre Analysis

```python
genreplot = (
    master
    .groupby("Genre")["Total_Signatures"]
    .mean()
    .sort_values()
)

genreplot.plot.bar(figsize=(14,6))
```
